CASTEO DE VARIABLES

```python
Cantidad = "3"
print(type(Cantidad))

Cantidad = int(cantidad)
print(type(Cantidad))
```

```python
precio = "130.5"
print(type(precio))

precio = float(precio)
print(type(precio))
print(precio)

precio = int(precio)
print(precio)
```

```python
activo = -1000
print(activo)
activo = bool(activo)
print(activo)
```

```python
precio = "3.5"
precio = bool(precio)

print(precio)
precio = int(precio)
print(precio)
```
